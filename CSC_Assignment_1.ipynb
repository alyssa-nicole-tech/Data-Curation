## Part 1: How bad is the flood? (Text Statistics)

Before we start coding solutions, we need to know the scale of the problem.
In NLP, we distinguish between:
*   **Tokens**: Every single word count (The total length of the inbox).
*   **Types**: The number of *unique* words (The vocabulary size).

If the Type/Token ratio is low, students are repeating the same words over and over (good for us!).

```python
def print_stats(text):

  tokens = nltk.word_tokenize(text)
  types = set(tokens)
  print(f"Total Tokens: {len(tokens)}")
  print(f"Unique Types: {len(types)}")
  return tokens

tokens = print_stats(raw_data)
```

### Removing the Noise (Stopwords)

You probably noticed that words like "is", "the", and "for" are appearing a lot. These are **Stopwords**.
They don't tell us what the student actually wants.
Let's remove them to see the *actual* vocabulary of the students.

```python
import nltk
nltk.download('stopwords')
```

```python
# Task 1.2: Remove Stopwords
# 1. Import stopwords from nltk.corpus
from nltk.corpus import stopwords
# 2. Filter your 'tokens' list to remove any word that is in the English stopwords list.
# 3. Print the new Type/Token count.

def remove_noise(tokens):
    # TODO: Import and use stopwords
    stop_words = (stopwords.words("english"))
    clean_tokens = []

    for token in tokens:
        if token.lower() not in stop_words:
            clean_tokens.append(token)

    print(f"Cleaned Tokens: {len(clean_tokens)}")
    print(f"Cleaned Types: {len(set(clean_tokens))}")
    return clean_tokens

clean_tokens = remove_noise(tokens)
```

## Part 2: Using Regex

Okay, we have a cleaner list. Now we need to answer specific questions.

### Question 1: How many students are asking about the EXAM?
We need to find every mention of the exam.
Students shout "EXAM", write "Exam", or just "exam".
If we miss one, we cannot accurately gauge the demand.

```python
def count_exams(text):
    matches = re.findall(r"\bexam\b", text, re.IGNORECASE)
    return len(matches)

print(f"Exam mentions: {count_exams(raw_data)}")
```

### Question 2: How many students are asking about GRADES?

This is harder. It's not just capitalization.
Students write:
*   "Grade"
*   "grading"
*   "graded"

Try to catch them all with one complex regex.

```python

def count_grades_regex(text):
    matches = re.findall(r"\bgrad(?:e|ing|ed)\b", text, re.IGNORECASE)
    return len(matches)

print(f"Grade mentions (Regex): {count_grades_regex(raw_data)}")
```

## Part 4: Who are they? (Structured Extraction)

We know *what* they want. Now we need to know *who* they are and *which class* they are in.

This is where Regex shines: **Structured Extraction**.
We need to pull out specific entities like Email Addresses and Course Numbers.

```python
def get_emails(text):
    emails = re.findall(r"[\w.-]+@[\w.-]+", text)
    return emails

print(f"Senders: {get_emails(raw_data)}")
```

```python
def get_course_counts(text):
    numbers = re.findall(r"CSC[\s-]?([0-9]{3})", text)
    return Counter(numbers)

print(f"Class Counts: {get_course_counts(raw_data)}")
```

## Part 5: The Auto-Reply Bot

It is time. We have the content analysis. Let's automate the replies.

We will define a **Knowledge Base** (a dictionary) that maps Intents to Regex Patterns.
Then we loop through the email subjects and fire the correct reply based on the keyword match.

```python
intents = {
    "Exam": r"exam|test|quiz|midterm",
    "Grade": r"grade|grading|graded"
}

responses = {
    "Exam": "The exam is on the 20th. Study hard.",
    "Grade": "Grades will be posted soon."
}

# 2. Process Emails
print("--- Auto-Reply Log ---")

for line in raw_data.split('\n'):
    if line.startswith("Subject:"):
        found_intent = False

        for intent, pattern in intents.items():
            if re.search(pattern, line, re.IGNORECASE):
                print(responses[intent])
                found_intent = True
                break

        if not found_intent:
            print("No automatic response available.")
```
